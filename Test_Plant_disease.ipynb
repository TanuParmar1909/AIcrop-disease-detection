```python
import cv2
import os

# Let's set a default image path from the downloaded Kaggle dataset to preview
# The dataset is downloaded to '/kaggle/input/plantvillage-dataset'
base_path = '/kaggle/input/plantvillage-dataset'

# Find the first available image in the dataset to visualize
image_path = None
for root, dirs, files in os.walk(base_path):
    for file in files:
        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
            image_path = os.path.join(root, file)
            break
    if image_path:
        break

print("Selected Image Path:", image_path)
if image_path:
    img = cv2.imread(image_path)
    print("Image shape:", img.shape)
```

```python
import cv2
image_path='/content/Screenshot 2026-11-12 214630.png'
img= cv2.imread(image_path)
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)     #colour bgr colour to rgb
plt.imshow(img)
plt.title('Test Image')
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
image=tf.keras.preprocessing.image.load_img(image_path,target_size=(128,128))
input_arr=tf.keras.preprocessing.image.img_to_array(image)
input_arr=np.array([input_arr])     #convert single into batch
print(input_arr.shape)
```

```python
prediction= model.predict(input_arr)
prediction.prediction.shape
```

```python
result_index=np.argmax(prediction)
result_index
```
